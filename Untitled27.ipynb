```python
import re

pattern = input("Введіть регулярний вираз: ")

try:
    compiled_pattern = re.compile(pattern)
    print("Регулярний вираз успішно скомпільовано!")
except re.error as e:
    print(f"Помилка: некоректний регулярний вираз. Деталі: {e}")
```

```python
import re

pattern = input("Введіть регулярний вираз: ")

try:
compiled_pattern = re.compile(pattern)
print("Регулярний вираз успішно скомпільовано!")

text = input("Введіть текст: ")
# Додатково можна перевірити збіг, якщо потрібно
if compiled_pattern.search(text):
    print("Знайдено збіг у тексті!")
else:
    print("Збігів у тексті не знайдено.")
except re.error:
print("Помилка: регулярний вираз має некоректний синтаксис.")
```

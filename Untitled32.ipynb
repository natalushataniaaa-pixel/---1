```python
import os
import re
from collections import Counter
import unittest

class EmptyTextError(Exception):
pass

def analyze_text_file(filename):
try:
with open(filename, "r", encoding="utf-8") as file:
text = file.read()
except FileNotFoundError:
raise FileNotFoundError(f"Файл '{filename}' не знайдено.")
except PermissionError:
raise PermissionError(f"Недостатньо прав доступу до файлу '{filename}'.")

words = re.findall(r"\b[а-яА-Яа-щьюяєіїґА-ЩЬЯЄІЇҐa-zA-Z]+\b", text.lower())

if not text.strip() or not words:
    raise EmptyTextError("Текст є порожнім або не містить слів.")

return len(words), dict(Counter(words))
class TestTextAnalyzer(unittest.TestCase):

def setUp(self):
    with open("test_sample.txt", "w", encoding="utf-8") as f:
        f.write("Привіт світ, це тестовий текст для перевірки слів. Hello world!")

def tearDown(self):
    for fname in ["test_sample.txt", "test_empty.txt"]:
        if os.path.exists(fname):
            os.remove(fname)

def test_word_count_and_frequency(self):
    count, freq = analyze_text_file("test_sample.txt")
    self.assertEqual(count, 10)
    self.assertIn("привіт", freq)
    self.assertEqual(freq["привіт"], 1)

def test_empty_text_exception(self):
    with open("test_empty.txt", "w", encoding="utf-8") as f:
        f.write("   ")
    with self.assertRaises(EmptyTextError):
        analyze_text_file("test_empty.txt")

def test_file_not_found_exception(self):
    with self.assertRaises(FileNotFoundError):
        analyze_text_file("non_existent_file.txt")
if name == "main":
unittest.main()
```
